```python
import pandas as pd
from IPython.display import display

df = pd.read_excel("vgsales.xlsx")
sale_cols = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]
df[sale_cols] = df[sale_cols].apply(pd.to_numeric, errors="coerce")
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df = df.dropna(how="any").copy()
df["Year"] = df["Year"].astype("int64")
df = df.sort_values("Global_Sales", ascending=False).reset_index(drop=True)

df.to_excel("vgsalesclean.xlsx", index=False)

numeric_cols = df.select_dtypes(include="number").columns
styled_df = (
    df.style
    .format({col: "{:.2f}" for col in sale_cols}, na_rep="")
    .set_properties(subset=numeric_cols, **{"text-align": "right"})
    .set_properties(subset=[c for c in df.columns if c not in numeric_cols], **{"text-align": "left"})
)
display(styled_df)
```
